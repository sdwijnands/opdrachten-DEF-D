## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
omega_0 = np.sqrt((100*(np.pi)**2)/4)
print(omega_0)
```

```python
# Voeg hier je plot in.
def f(w):
    return np.abs(1/(1-((w**2)/(omega_0**2))))

a= np.linspace(0,100,1000)

plt.figure()
plt.plot(a,f(a))
plt.xscale('log')
plt.yscale('log')
plt.ylabel('amplitude overdracht [-]')
plt.xlabel('frequentie [Hz]')

```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g= 9.81
dx=g/(omega_0**2)
print(dx)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
def C(m,a):
    return (m*g)/(a*dx)

m= np.linspace(0,10,100)

plt.figure()
plt.plot(m,C(m,1),c='b',label='1*dx')
plt.plot(m,C(m,2),c='r',label='2*dx')
plt.plot(m,C(m,5),c='g',label='5*dx')
plt.xlabel("m (kg)")
plt.ylabel("C (N/m)")
plt.legend()
plt.title("Ontwerpgrafiek massa-veersysteem")

C=C(0.015,1)
C
plt.savefig('opdracht_3.png')
```

```python
# Voeg hier je plot in.
def f(w):
    return np.abs(1/(1-((w**2)/(omega_0**2))))

def f2(w,x):
    return np.abs(1/(1-((w**2)/((g)/(x)))))

a= np.linspace(0,100,1000)

plt.figure()
plt.plot(a,f(a),label='ontwerp-eis')
plt.xscale('log')
plt.yscale('log')
plt.ylabel('amplitude overdracht [-]')
plt.xlabel('frequentie [Hz]')

plt.plot(a,f2(a,0.07),label='eigen ontwerp')
plt.legend()
plt.title("ontwerpgrafiek amplitude overdracht")
plt.savefig("ontwerpgrafiek_amplitudeoverdracht.png", dpi=450)

```

```python
# Voeg hier je plots in.
freq= np.array([5.2,5.5,6.6,5,4.4,3.6,3.6,3.3,3.2,1.5,2.7,2.9,3,2.5,2.7,3,2.9,3,2])
ada= np.array([0.25/1.28,0.2/1.41,0.34/3.08,0.29/3.63,0.43/1.08,0.18/0.61,0.24/0.85,0.63/1.5,1.11/3.66,0.2/0.11,0.29/0.65,0.22/0.24,0.28/0.12,0.13/0.07,0.29/0.65,0.28/0.12,0.41/0.22,1.03/0.53,2.23/0.46])

def f(w):
    return np.abs(1/(1-((w**2)/(omega_0**2))))

def f2(w,x):
    return np.abs(1/(1-((w**2)/((g)/(x)))))

a= np.linspace(0,100,1000)

plt.figure()
plt.plot(a,f(a),label='ontwerp-eis')
plt.xscale('log')
plt.yscale('log')
plt.ylabel('amplitude overdracht [-]')
plt.xlabel('frequentie [Hz]')


plt.plot(a,f2(a,0.07),label='eigen ontwerp')

plt.plot(freq,ada,'r.',label='meetpunten')

xerr= 0.2
yerr= 0.1

plt.errorbar(freq,ada,xerr=xerr,yerr=yerr,fmt='r.',label='onzekerheid')

plt.legend()
plt.savefig('opdracht7')
```

de meetpunten volgen wel dezelfde lijn als de functie, maar zijn een stuk naar links verschoven. dit komt omdat het elastiek dat wij hebben gebruikt in onze opstelling zich niet ideaal gedraagt volgens de wet van hooke.
